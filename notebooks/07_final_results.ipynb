
## Conclusions

- Tuned Random Forest achieved the highest
  cross-validation macro F1 of 72.14%.
- Baseline Logistic Regression achieved
  dropout recall of 76.76%.
- First-semester XGBoost achieved 71.41%
  accuracy and 67.61% dropout recall.
- First-semester information substantially
  improved performance over enrollment data.

## Breaking Educational Barriers

The models could support earlier academic
interventions and help advisors identify
students who may benefit from additional support.

## Limitations

The experiments are retrospective.
Feature availability, subgroup reliability,
probability calibration, and independent
validation require further investigation.

Predictions should support human decisions,
not determine access to educational opportunities.
