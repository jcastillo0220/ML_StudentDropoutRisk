```python
import pickle

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    classification_report,
    accuracy_score
)
```

```python
with open(
    "../results/processed_data.pkl",
    "rb"
) as file:

    data = pickle.load(file)
```

```python
rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42
)

rf.fit(
    data["X_train"],
    data["y_train"]
)
```

```python
rf_pred = rf.predict(
    data["X_test"]
)
```

```python
print(
    classification_report(
        data["y_test"],
        rf_pred
    )
)
```

```python
rf_accuracy = accuracy_score(
    data["y_test"],
    rf_pred
)

rf_accuracy
```

```python
import pandas as pd

importance = pd.DataFrame({
    "Feature":
        data["X_train"].columns,
    "Importance":
        rf.feature_importances_
})

importance.sort_values(
    "Importance",
    ascending=False
).head(15)
```
