## First-Semester Threshold Analysis

```

display(pd.read_csv(
    RESULTS / "threshold_analysis.csv"
).round(4))

display(Image(filename=str(
    CHARTS / "first_semester_precision_recall.png"
)))
```


## Interpretation

- Baseline Logistic Regression identified the
  most actual dropouts among the full-information
  multiclass models.
- Tuned Random Forest achieved the highest
  overall macro F1.
- Subgroup differences reveal potential gaps
  in intervention coverage.
- Small subgroup samples require caution.
- Threshold results are exploratory and should
  be validated before operational use.
